# v2 Agent — `searchFiles` + LLM synthesis

A single read-only agent that takes a request, **finds every resembling file, reads the
relevant part, and hands the evidence to a real LLM that answers in natural language.**

```
user request
     |
     v
[ intent ]   filenames + keywords from the raw request (regex, no model)
     |
     v
[ recall ]   find EVERY resembling file:  exact -> glob -> substring -> fuzzy>=0.6
     |         grep by content when no filename was named
     v
[ read ]     capped window, centred on the match
     |
     v
[ bundle ]   compact evidence, paths + line numbers, never a raw dump
     |
     v
[ LLM ]      Groq (OpenAI-compatible) -> natural-language answer   <-- NEW
     |
     v
user
```

**What changed in v2**
1. **Real LLM call.** `call_llm()` posts to any OpenAI-compatible `/chat/completions`.
   Defaults to Groq. Activates the moment `GROQ_API_KEY` is set; degrades to the raw
   bundle (never a crash) when it isn't.
2. **Exhaustive recall.** `find_files()` now returns *every* resembling file, ranked by
   match kind, instead of stopping at the first hit.

**Environment (measured, not assumed)**

| Want | Status here | Using |
|---|---|---|
| `fd` | not installed | `find` |
| `rg` | not installed | `grep -rIn` |
| `plocate` | installed, **stale index** | skipped on purpose |
| `--max-filesize` | not supported by grep 3.12 | aggressive `SKIP_DIRS` |
| Groq key | **not set** | set it, see the LLM cell |

**Not in this milestone:** planner, webSearch, executor, GLiNER, DSPy/ReAct.

````python
# ===== Cell: config =====
from __future__ import annotations

import difflib
import fnmatch
import json
import os
import re
import shutil
import subprocess
import time
from pathlib import Path

# --- the jail. Every filesystem path must resolve inside one of these.
ROOTS = [Path("/home/princ3kr")]

# --- directories pruned from every recursive walk. Mostly app data, not user code.
SKIP_DIRS = {
    ".git", "node_modules", ".venv", "venv", "__pycache__",
    ".mypy_cache", ".pytest_cache", ".ipynb_checkpoints", "dist", "build",
    ".vscode", "site-packages",
    ".cache", ".config", ".cargo", ".dotnet", ".nv", ".npm", ".snap", ".var",
    ".triton", ".gnupg", ".ipython", ".jupyter", ".icons", ".local",
    ".mozilla", ".thunderbird", "scikit_learn_data",
}

# --- never opened as text (no extractors installed for these anyway)
SKIP_EXT = {
    ".pyc", ".so", ".dylib", ".dll", ".png", ".jpg", ".jpeg", ".gif", ".webp",
    ".ico", ".pdf", ".docx", ".doc", ".xlsx", ".zip", ".gz", ".tar", ".bz2",
    ".xz", ".7z", ".whl", ".mp4", ".mov", ".mp3", ".wav", ".woff", ".woff2",
    ".ttf", ".eot", ".db", ".sqlite", ".bin", ".exe", ".o", ".a",
}

# --- limits
FIND_MAX_RESULTS  = 200     # v2: was 20. "find EVERY resembling file"
FUZZY_POOL_CAP    = 4_000   # candidates considered for fuzzy ranking
FUZZY_CUTOFF      = 0.60    # difflib ratio below which a file is NOT "resembling"
MAX_RESULTS       = 20      # content-search hits per keyword
MAX_FILE_BYTES    = 50_000
MAX_LINES         = 200
GREP_MAX_PER_FILE = 3
READ_CONTEXT      = 25
BINARY_SNIFF      = 8_192
TIMEOUT_S         = 15

# KNOWN LIMITATION: this grep has no --max-filesize, so a large *text* file in an
# unpruned directory is still scanned. read_file() caps size; search_content() cannot.
# Hence the aggressive SKIP_DIRS. Real fix = size-aware walk or an index.

TOOLS = ("find", "grep", "read")
AGENT_NAME = "searchFiles"

print("fd    :", shutil.which("fd") or "not installed -> using find")
print("rg    :", shutil.which("rg") or "not installed -> using grep")
print("locate:", shutil.which("locate") or "-", "(skipped: stale index)")
print("jail  :", [str(r) for r in ROOTS])
````

## Logging

Plain `list[dict]`, no database. One record per call — including the LLM call, with token
usage. Same fields the full harness will push to SQLite, so swapping `LOGS.append()` for
an insert should not change any call site.

````python
# ===== Cell: logger =====
LOGS: list[dict] = []


def log(action: str, *, detail: str = "", status: str = "ok",
        t0: float | None = None, **extra) -> dict:
    rec = {
        "n": len(LOGS) + 1,
        "agent": AGENT_NAME,
        "action": action,
        "detail": detail,
        "status": status,
        "duration_ms": round((time.perf_counter() - t0) * 1000, 2) if t0 else None,
        "ts": time.strftime("%H:%M:%S"),
    }
    rec.update(extra)
    LOGS.append(rec)
    return rec


def show_logs(records: list[dict] | None = None) -> None:
    rows = LOGS if records is None else records
    if not rows:
        print("LOGS is empty — run ask(...) first.")
        return
    w = max(len(str(r["action"])) for r in rows) + 2
    print(f"{'#':>3}  {'time':<9} {'action':<{w}} {'status':<9} {'ms':>8}  detail")
    print("-" * (48 + w))
    for r in rows:
        ms = f"{r['duration_ms']:.1f}" if r["duration_ms"] is not None else "-"
        flag = "" if r["status"] == "ok" else "  <-- "
        print(f"{r['n']:>3}  {r['ts']:<9} {str(r['action']):<{w}} {r['status']:<9} "
              f"{ms:>8}  {str(r['detail'])[:66]}{flag}")
    ok = sum(1 for r in rows if r["status"] == "ok")
    print("-" * (48 + w))
    print(f"{len(rows)} call(s) | {ok} ok | {len(rows) - ok} not-ok")


def clear_logs() -> None:
    LOGS.clear()
    print("LOGS cleared.")
````

## The jail — `safe_resolve`

The security model, and it is **code, not a prompt instruction**.

1. **Containment** — resolved path must sit inside `ROOTS`. `.resolve()` runs *before* the
   check, so `..` traversal and escaping symlinks are both caught.
2. **Read-only** — only three tools exist; files are opened in `"rb"` and nothing else.
3. **Allowlist** — `call_tool()` dispatches by name and fails closed on unknown names.

````python
# ===== Cell: path jail + tool allowlist =====
ROOTS = [r.resolve() for r in ROOTS]


def safe_resolve(path) -> Path:
    """Resolve `path`, guaranteeing it lives inside ROOTS. Raises PermissionError otherwise."""
    p = Path(str(path)).expanduser()
    candidates = [p] if p.is_absolute() else [Path.cwd() / p] + [r / p for r in ROOTS]
    last_err = None
    for cand in candidates:
        try:
            rp = cand.resolve()
        except (OSError, RuntimeError) as e:
            last_err = e
            continue
        if any(rp == r or r in rp.parents for r in ROOTS):
            return rp
        last_err = None
    raise PermissionError(
        f"outside jail: {path!r} -> {p.absolute()} is not inside {[str(r) for r in ROOTS]}"
    ) from last_err


def call_tool(name: str, **kwargs):
    """Dispatch by allowlist. An invented tool name errors instead of executing."""
    if name not in TOOLS:
        raise ValueError(f"tool {name!r} not allowed; allowed = {list(TOOLS)}")
    return {"find": find_files, "grep": search_content, "read": read_file}[name](**kwargs)


for bad in ["/etc/passwd", "/", "/home/princ3kr/Desktop/../../../etc/shadow"]:
    try:
        safe_resolve(bad)
        print(f"  LEAK  {bad}")
    except PermissionError:
        print(f"  blocked  {bad}")
print("  allowed  ~/Desktop/learnings/glinerTest.ipynb ->",
      safe_resolve("~/Desktop/learnings/glinerTest.ipynb").name)
````

## Tool 1 — `find_files` with exhaustive similarity

The v2 change. "Find every file which resembles it" means matching on **four tiers**,
in rank order:

| rank | kind | `utils.py` matches |
|---|---|---|
| 0 | `exact` | `utils.py` |
| 1 | `glob` | `ut*.py` |
| 2 | `substring` | `utils_test.py`, `my_utils.py` |
| 3 | `fuzzy` | `util.py` (difflib 0.89) |
| — | rejected | `unrelated.py` (0.38 < 0.60) |

Two passes:
- **pass 1** — `find -iname *pattern*` (substring, fast, filesystem-native)
- **pass 2** — only if pass 1 came back short: gather a candidate pool (same extension if
  the pattern has one, else all files, capped) and rank with `difflib` on the stem

`fuzzy=False` disables pass 2 and gives you the plain substring behaviour.

````python
# ===== Cell: tool 1 — find EVERY resembling file =====
def _stem(name: str) -> str:
    """basename without extension, lowercased -- what similarity is judged on."""
    return Path(name).stem.lower()


def _match_kind(pattern: str, path) -> tuple[int, float, str] | tuple[None, float, None]:
    """Rank one candidate. Returns (rank, score, kind) or (None, ratio, None) if unlike."""
    name = Path(path).name
    low, pl = name.lower(), pattern.lower()
    ps, ns = _stem(pl), _stem(low)
    if low == pl:
        return 0, 1.0, "exact"
    if fnmatch.fnmatch(low, pl):
        return 1, 0.95, "glob"
    if pl and (pl in low or (ps and ps in ns)):
        return 2, 0.80, "substring"
    ratio = difflib.SequenceMatcher(None, ns, ps).ratio() if (ns and ps) else 0.0
    if ratio >= FUZZY_CUTOFF:
        return 3, round(ratio, 3), "fuzzy"
    return None, round(ratio, 3), None


def _find_raw(pattern: str | None, base: Path, limit: int | None = None) -> list[str]:
    """Raw `find` output, pruned. pattern=None means every file."""
    cmd = ["find", str(base), "("]
    for i, d in enumerate(sorted(SKIP_DIRS)):
        if i:
            cmd.append("-o")
        cmd += ["-name", d]
    cmd += [")", "-prune", "-o", "-type", "f", "-print"]
    if pattern:
        cmd += ["-iname", pattern]
    try:
        out = subprocess.run(cmd, capture_output=True, text=True,
                             timeout=TIMEOUT_S, check=False)
    except subprocess.TimeoutExpired:
        return []
    lines = [l for l in out.stdout.splitlines() if l.strip()]
    return lines[:limit] if limit else lines


def find_files(pattern: str, root=None, max_results: int = FIND_MAX_RESULTS,
               fuzzy: bool = True) -> list[dict]:
    """Every file resembling `pattern`, ranked best-first.

    Returns [{path, rank, kind, score}, ...] ranked by kind then score.
    """
    t0 = time.perf_counter()
    base = safe_resolve(root or ROOTS[0])

    if not any(ch in pattern for ch in "*?["):
        pattern_wild = f"*{pattern}*"
    else:
        pattern_wild = pattern

    best: dict[Path, dict] = {}

    def consider(raw: str):
        try:
            rp = safe_resolve(raw)          # never trust raw find output
        except PermissionError:
            return
        if rp in best:
            return
        # binary/unextractable files are not useful "similar source files"
        if rp.suffix.lower() in SKIP_EXT:
            return
        rank, score, kind = _match_kind(pattern, rp)
        if kind:
            best[rp] = {"path": rp, "rank": rank, "kind": kind, "score": score}

    for raw in _find_raw(pattern_wild, base):
        consider(raw)

    # pass 2: broaden only if recall is still short
    pool_truncated = False
    if fuzzy and len(best) < max_results:
        ext = Path(pattern).suffix
        pool = _find_raw(f"*{ext}" if ext else None, base, limit=FUZZY_POOL_CAP + 1)
        pool_truncated = len(pool) > FUZZY_POOL_CAP
        for raw in pool[:FUZZY_POOL_CAP]:
            consider(raw)

    out = sorted(best.values(), key=lambda d: (d["rank"], -d["score"], str(d["path"])))
    out = out[:max_results]
    kinds = {}
    for d in out:
        kinds[d["kind"]] = kinds.get(d["kind"], 0) + 1

    log("find", detail=f"{pattern} -> {len(out)} similar",
        status="ok" if out else "not_found", t0=t0,
        found=len(out), by_kind=kinds, fuzzy_pass2=fuzzy,
        pool_truncated=pool_truncated, tool="find")
    return out
````

## Tool 2 — `search_content`

`grep -rIn` — recursive, line numbers, skips binary. Literal (`-F`) by default so a query
containing `(`, `|`, or `.` can't corrupt the pattern; `regex=True` switches to `-E`.
Returns compact `(path, line, text)` triples.

````python
# ===== Cell: tool 2 — grep inside files =====
def search_content(query: str, root=None, max_results: int = MAX_RESULTS,
                   regex: bool = False,
                   max_per_file: int = GREP_MAX_PER_FILE) -> list[dict]:
    t0 = time.perf_counter()
    base = safe_resolve(root or ROOTS[0])

    cmd = ["grep", "-rI", "-n", "-m", str(max_per_file)]
    cmd += ["-E" if regex else "-F", query]
    for d in sorted(SKIP_DIRS):
        cmd.append(f"--exclude-dir={d}")
    for e in sorted(SKIP_EXT):
        cmd.append(f"--exclude=*{e}")
    cmd.append(str(base))

    try:
        out = subprocess.run(cmd, capture_output=True, text=True,
                             timeout=TIMEOUT_S, check=False)
    except subprocess.TimeoutExpired:
        log("grep", detail=f"{query!r} (TIMEOUT {TIMEOUT_S}s)", status="timeout", t0=t0)
        return []

    hits, seen = [], {}
    for line in out.stdout.splitlines():
        if len(hits) >= max_results:
            break
        if ":" not in line:
            continue
        raw_path, _, rest = line.partition(":")
        parts = rest.split(":", 1)
        if len(parts) != 2 or not parts[0].isdigit():
            continue
        line_no, body = int(parts[0]), parts[1]
        if seen.get(raw_path, 0) >= max_per_file:
            continue
        try:
            p = safe_resolve(raw_path)
        except PermissionError:
            continue
        seen[raw_path] = seen.get(raw_path, 0) + 1
        hits.append({"path": p, "line_no": line_no, "text": body.strip()[:300]})

    log("grep", detail=f"{query!r} -> {len(hits)} hit(s)",
        status="ok" if hits else "not_found", t0=t0,
        found=len(hits), files=len(seen), tool="grep")
    return hits
````

## Tool 3 — `read_file`

Disk isn't the bottleneck; *limiting what comes back* is.
- hard byte cap and line cap
- NUL-byte sniff on the first 8 KB, so binaries are reported, not spewed
- extensions with no extractor installed are refused up front
- accepts a line range, so the agent reads the window around a hit, not the whole file

````python
# ===== Cell: tool 3 — read with limits =====
def read_file(path, start_line: int = 1, max_lines: int = MAX_LINES,
              max_bytes: int = MAX_FILE_BYTES) -> dict:
    """Capped read. Never raises for ordinary problems -- returns status='skipped'/
    'binary'/'not_found' with a reason, so the agent can say 'not found' not guess."""
    t0 = time.perf_counter()
    try:
        p = safe_resolve(path)
    except PermissionError as e:
        log("read", detail=str(path), status="blocked", t0=t0, tool="read")
        return {"path": str(path), "status": "blocked", "reason": str(e), "text": ""}

    if not p.is_file():
        log("read", detail=str(p), status="not_found", t0=t0, tool="read")
        return {"path": str(p), "status": "not_found", "reason": "not a file", "text": ""}

    if p.suffix.lower() in SKIP_EXT:
        log("read", detail=f"{p.name} ({p.suffix})", status="skipped", t0=t0, tool="read")
        return {"path": str(p), "status": "skipped",
                "reason": f"{p.suffix} needs an extractor (none installed)", "text": ""}

    size = p.stat().st_size
    if size > max_bytes * 4:
        log("read", detail=f"{p.name} {size}B > cap", status="skipped", t0=t0, tool="read")
        return {"path": str(p), "status": "skipped",
                "reason": f"file is {size:,}B, too large to read blindly", "text": ""}

    try:
        with open(p, "rb") as fh:                     # read-only, always
            head = fh.read(BINARY_SNIFF)
            if b"\x00" in head:
                log("read", detail=f"{p.name} (binary)", status="binary", t0=t0, tool="read")
                return {"path": str(p), "status": "binary",
                        "reason": "contains NUL bytes", "text": ""}
            fh.seek(0)
            raw = fh.read(max_bytes)
    except OSError as e:
        log("read", detail=f"{p.name} ({e.strerror})", status="error", t0=t0, tool="read")
        return {"path": str(p), "status": "error", "reason": str(e), "text": ""}

    lines = raw.decode("utf-8", errors="replace").splitlines()
    start = max(1, int(start_line))
    window = lines[start - 1: start - 1 + max_lines]
    shown = "\n".join(f"{start + i:>5} | {ln}" for i, ln in enumerate(window))

    log("read", detail=f"{p.name} L{start}-L{start + len(window) - 1}",
        status="ok", t0=t0, lines=len(window), bytes=len(raw), tool="read")
    return {"path": str(p), "status": "ok", "start_line": start,
            "n_lines": len(window), "total_lines": len(lines),
            "truncated": len(window) < (len(lines) - start + 1), "text": shown}
````

## The agent

1. **intent** — filenames + keywords from the raw request (regex, no model)
2. **recall** — if filenames were named, union results across **all** of them (no early
   break), each returning every resembling file. Otherwise grep by content: one keyword
   normally, all of them when `exhaustive=True`.
3. **read** — window centred on the match
4. **evidence** — compact, citable

The `find vs grep` split is a hardcoded heuristic, not a model. That is deliberate: it's
the piece DSPy replaces later, and a dumb deterministic version is easier to debug.

````python
# ===== Cell: the agent loop =====
FILENAME_RE = re.compile(
    r"[\w\-./~]*[\w\-]+\.(?:py|ipynb|md|txt|json|ya?ml|toml|cfg|ini|js|jsx|ts|tsx|"
    r"html|css|sh|bash|sql|csv|log|env|lock|c|h|cpp|java|go|rs|rb|php)", re.I)

STOPWORDS = {
    "the", "a", "an", "is", "are", "was", "in", "on", "at", "to", "of", "and", "or",
    "for", "with", "why", "how", "what", "where", "which", "does", "do", "did", "can",
    "you", "i", "me", "my", "it", "this", "that", "there", "here", "be", "been", "not",
    "but", "if", "then", "so", "about", "from", "by", "please", "help", "show", "find",
    "tell", "explain", "look", "check", "file", "files", "code", "use", "using", "get",
}

EXTRACTORS = {".pdf": "pypdf", ".docx": "python-docx"}


def extract_intent(query: str) -> dict:
    """Filenames and keywords from the raw request. Cheap regex, no model."""
    filenames, seen = [], set()
    for m in FILENAME_RE.finditer(query):
        fn = m.group(0).strip("./")
        if fn.lower() not in seen:
            seen.add(fn.lower())
            filenames.append(fn)
    keywords = [w for w in re.findall(r"[A-Za-z_][A-Za-z0-9_.]{2,}", query)
                if w.lower() not in STOPWORDS and not w.endswith(".")]
    return {"filenames": filenames, "keywords": list(dict.fromkeys(keywords))[:5],
            "raw": query,
            "needs_extractor": any(query.lower().endswith(e) for e in EXTRACTORS)}


def run_agent(user_query: str, top_files: int = 3, root=None,
              exhaustive: bool | None = None) -> dict:
    """intent -> recall -> read -> evidence.

    exhaustive=None auto-enables when the request names a file, because
    "find this file" means "every file resembling it".
    """
    t_start = time.perf_counter()
    intent = extract_intent(user_query)
    if exhaustive is None:
        exhaustive = bool(intent["filenames"])

    ranked: list[dict] = []
    seen_paths: set[Path] = set()
    searches: list[dict] = []

    # 1. name-based recall: ALL filename candidates, union, no early break
    for fn in intent["filenames"]:
        found = call_tool("find", pattern=fn, root=root, max_results=FIND_MAX_RESULTS)
        searches.append({"by": "find", "query": fn, "n": len(found),
                         "kinds": {k: sum(1 for f in found if f["kind"] == k)
                                   for k in ("exact", "glob", "substring", "fuzzy")
                                   if any(f["kind"] == k for f in found)}})
        for h in found:
            if h["path"] not in seen_paths:
                seen_paths.add(h["path"])
                ranked.append(h)

    # 2. content-based recall: only if no filename matched, or when forced exhaustive
    if not ranked:
        for kw in (intent["keywords"] if exhaustive else intent["keywords"][:1]):
            found = call_tool("grep", query=kw, root=root, max_results=MAX_RESULTS)
            searches.append({"by": "grep", "query": kw, "n": len(found)})
            for h in found:
                if h["path"] not in seen_paths:
                    seen_paths.add(h["path"])
                    ranked.append({"path": h["path"], "rank": 9, "kind": "content",
                                   "score": 0.0, "line_no": h["line_no"],
                                   "text": h["text"]})

    if not ranked:
        log("run_agent", detail=user_query[:60], status="not_found", t0=t_start,
            searches=len(searches), tool="agent")
        return {"status": "not_found", "query": user_query, "intent": intent,
                "searches": searches, "evidence": [],
                "similar_files": [], "summary": "No file matched.",
                "confidence": 0.0, "elapsed_ms": round((time.perf_counter() - t_start) * 1000, 2)}

    # 3. read the top N, centred on the match
    evidence = []
    for h in ranked[:top_files]:
        line = h.get("line_no")
        r = call_tool("read", path=h["path"],
                      start_line=(line - READ_CONTEXT if line else 1))
        evidence.append({
            "path": str(h["path"]), "kind": h["kind"], "score": h["score"],
            "match_line": line, "match_text": h.get("text"),
            "read_status": r["status"], "truncated": r.get("truncated"),
            "excerpt": r.get("text", "")[:6000],
        })

    similar = [{"path": str(h["path"]), "kind": h["kind"], "score": h["score"]}
               for h in ranked]
    by_grep = any(s["by"] == "grep" for s in searches)
    rec = log("run_agent", detail=user_query[:60], status="ok", t0=t_start,
              files=len(evidence), similar=len(similar),
              searches=len(searches), tool="agent")
    return {
        "status": "ok", "query": user_query, "intent": intent,
        "searches": searches, "evidence": evidence, "similar_files": similar,
        "summary": f"Found {len(similar)} similar file(s); read {len(evidence)}.",
        "confidence": 0.8 if not by_grep else 0.6, "elapsed_ms": rec["duration_ms"],
    }
````

## The prompt bundle

What the LLM receives. Compact and citable — paths plus line numbers, never a raw dump.

It also carries the **full list of resembling files**, not just the ones that got read, so
the answer can say "there are 7 similar files, here they are" instead of implying it read
everything.

And the agent is allowed to say **"not found"** — `confidence` is explicit so the LLM
hedges rather than invents.

````python
# ===== Cell: prompt bundle =====
def build_prompt_bundle(result: dict) -> str:
    if result["status"] == "not_found":
        return (f"AGENT: {AGENT_NAME}\nQUERY: {result['query']}\nRESULT: not_found\n"
                f"SEARCHES ATTEMPTED: {result['searches']}\n"
                f"INSTRUCTION: Tell the user plainly that nothing matched. "
                f"Do NOT guess a filename or invent content.")

    q = result["query"]
    out = [f"AGENT: {AGENT_NAME} (read-only)", f"QUERY: {q}",
           f"SEARCHES: {result['searches']}"]

    sim = result.get("similar_files") or []
    if len(sim) > len(result["evidence"]):
        out.append(f"ALL {len(sim)} SIMILAR FILES FOUND (only the first "
                   f"{len(result['evidence'])} were read):")
        for s in sim:
            out.append(f"    [{s['kind']:<9} {s['score']}] {s['path']}")

    out += ["", f"EVIDENCE ({len(result['evidence'])} file(s) read):", ""]
    for i, e in enumerate(result["evidence"], 1):
        out.append(f"--- [{i}] {e['path']}   (matched by {e['kind']})")
        if e["match_line"]:
            out.append(f"    line {e['match_line']}: {e['match_text']}")
        if e["read_status"] != "ok":
            out.append(f"    (not read: {e['read_status']})")
            continue
        if e["truncated"]:
            out.append("    (excerpt, truncated)")
        out.append(e["excerpt"])
        out.append("")

    out += ["=" * 60,
            f"CONFIDENCE: {result['confidence']}   ({result['elapsed_ms']}ms)",
            "INSTRUCTION: Answer the query in natural language using ONLY the evidence "
            "above. Cite files as path:line. If the evidence does not answer it, say so "
            "plainly rather than guessing."]
    return "\n".join(out)
````

## The LLM

Any **OpenAI-compatible** `/chat/completions` endpoint — so Groq, OpenAI, OpenRouter,
Together, or a local Ollama all work by changing two env vars.

### Enable it

```bash
export GROQ_API_KEY="gsk_..."          # or LLM_API_KEY
export LLM_MODEL="llama-3.3-70b-versatile"
```

Groq hosts several free-tier models that work well for this: `llama-3.3-70b-versatile`
(default here), `openai/gpt-oss-120b`, `qwen/qwen3-32b`, `meta-llama/llama-4-scout-17b`.

### Design notes

- **`temperature=0.0`** — this is a retrieval task; creativity is a liability.
- **System prompt** pins the model to the evidence and forbids invention.
- **Never crashes.** No key or a network failure returns the bundle plus an instruction,
  and logs `status='no_key'` / `'error'` — the harness degrades, it doesn't die.
- Token usage is logged per call, so you can see the cost of the loop.

````python
# ===== Cell: LLM client (OpenAI-compatible) =====
import requests

LLM_BASE_URL = "https://api.groq.com/openai/v1"
LLM_API_KEY  = "gsk_<REDACTED>"
LLM_MODEL    = "llama-3.3-70b-versatile"
LLM_TIMEOUT  = 60
LLM_TEMPERATURE = 0.0

# Other providers, uncomment one:
# LLM_BASE_URL, LLM_MODEL = "https://api.openai.com/v1", "gpt-4o-mini"
# LLM_BASE_URL, LLM_MODEL = "https://openrouter.ai/api/v1", "anthropic/claude-3.5-sonnet"
# LLM_BASE_URL, LLM_MODEL = "http://localhost:11434/v1", "llama3.2"   # Ollama, no key

SYSTEM_PROMPT = """You are a precise code assistant answering questions about a codebase.

Rules:
- Use ONLY the evidence provided. Never invent file names, line numbers, or code.
- Cite sources inline as path/to/file.ext:12.
- If the evidence does not answer the question, say so plainly and name what is missing.
- If a file could not be read, say so rather than guessing its contents.
- When several similar files were found, list them and explain the difference.
- Be concise and concrete. No preamble, no restating the question."""


class LLMUnavailable(RuntimeError):
    pass


def llm_status() -> dict:
    key = "GROQ_API_KEY" if LLM_API_KEY.startswith("gsk_") else "LLM_API_KEY"
    return {
        "configured": bool(LLM_API_KEY),
        "base_url": LLM_BASE_URL,
        "model": LLM_MODEL,
        "hint": (f'export {key}="..."' if not LLM_API_KEY
                 else f"using {LLM_MODEL} @ {LLM_BASE_URL}"),
    }


def build_llm_messages(result: dict) -> list[dict]:
    return [{"role": "system", "content": SYSTEM_PROMPT},
            {"role": "user", "content": build_prompt_bundle(result)}]


def call_llm(messages: list[dict], *, temperature: float = LLM_TEMPERATURE,
             max_tokens: int = 900) -> dict:
    """POST to any OpenAI-compatible /chat/completions. Raises LLMUnavailable on failure."""
    if not LLM_API_KEY:
        raise LLMUnavailable("no API key set (GROQ_API_KEY / LLM_API_KEY)")

    t0 = time.perf_counter()
    url = f"{LLM_BASE_URL}/chat/completions"
    payload = {"model": LLM_MODEL, "messages": messages,
               "temperature": temperature, "max_tokens": max_tokens}
    headers = {"Authorization": f"Bearer {LLM_API_KEY}",
               "Content-Type": "application/json"}

    try:
        r = requests.post(url, headers=headers, json=payload, timeout=LLM_TIMEOUT)
    except requests.RequestException as e:
        log("llm", detail=f"{LLM_MODEL} network error", status="error", t0=t0,
            model=LLM_MODEL)
        raise LLMUnavailable(f"network error: {type(e).__name__}: {e}") from e

    if r.status_code != 200:
        log("llm", detail=f"HTTP {r.status_code}", status="error", t0=t0, model=LLM_MODEL)
        raise LLMUnavailable(f"HTTP {r.status_code}: {r.text[:300]}")

    data = r.json()
    try:
        text = data["choices"][0]["message"]["content"].strip()
    except (KeyError, IndexError, TypeError) as e:
        log("llm", detail="malformed response", status="error", t0=t0, model=LLM_MODEL)
        raise LLMUnavailable(f"unexpected response shape: {str(data)[:200]}") from e

    usage = data.get("usage", {}) or {}
    log("llm", detail=f"{data.get('model', LLM_MODEL)} -> {len(text)} chars",
        status="ok", t0=t0, model=data.get("model", LLM_MODEL),
        tokens=usage.get("total_tokens"), tool="llm")
    return {"text": text, "usage": usage, "model": data.get("model", LLM_MODEL)}


st = llm_status()
print("LLM base_url :", st["base_url"])
print("LLM model    :", st["model"])
print("LLM key      :", "SET" if st["configured"] else "NOT SET  -> export GROQ_API_KEY=\"gsk_...\"")
print("status       :", "LIVE" if st["configured"] else "offline (bundle fallback active)")
````

## `ask()` — the full loop

`ask()` runs the agent **and** the LLM. It returns natural language when a key is set, and
falls back to the raw bundle plus a copy-pasteable instruction when it isn't. It never
raises for a missing key or a network blip.

````python
# ===== Cell: ask() =====
NOT_FOUND_REPLY = ("I couldn't find anything resembling that. I searched by filename and "
                   "by file contents, and got no matches — so rather than guess, here's "
                   "what I looked for.")


def synthesize(result: dict) -> dict | None:
    """Evidence -> natural language. Returns None (and logs why) if the LLM is unusable."""
    try:
        return call_llm(build_llm_messages(result))
    except LLMUnavailable as e:
        log("synthesize", detail=str(e)[:80],
            status="no_key" if not LLM_API_KEY else "error", tool="llm")
        return None


def ask(user_query: str, top_files: int = 3, root=None,
        exhaustive: bool | None = None) -> str | None:
    """plan -> recall -> read -> LLM -> natural language answer."""
    result = run_agent(user_query, top_files=top_files, root=root, exhaustive=exhaustive)

    print("=" * 62)
    print(f"AGENT: {result['status']} | similar={len(result.get('similar_files', []))} "
          f"read={len(result['evidence'])} | {result['elapsed_ms']}ms")
    print("=" * 62)

    out = synthesize(result)

    if out:
        print("\n--- ANSWER ---\n")
        print(out["text"])
        print(f"\n[{out['model']} | {out['usage'].get('total_tokens', '?')} tokens | "
              f"conf {result['confidence']}]")
        return out["text"]

    # ---- graceful degradation, never a crash
    if result["status"] == "not_found":
        print("\n--- ANSWER (offline) ---\n")
        print(NOT_FOUND_REPLY)
        return NOT_FOUND_REPLY

    print("\n--- LLM NOT AVAILABLE: raw evidence instead ---")
    print(build_prompt_bundle(result))
    print(f"\n>> set a key to get natural language:  {llm_status()['hint']}")
    print(">> or paste the block above to any LLM.")
    return None
````

## Smoke tests

These check the harness, not the agent's answers. Notably **#9** asserts the similarity
logic precisely: given `utils.py`, `utils_test.py`, `my_utils.py`, `util.py`,
`unrelated.py`, it must return the first four and reject the last.

**#11 stubs the HTTP transport**, so the full loop — bundle built, request shaped,
natural-language answer parsed and returned — is verified without an API key.

````python
# ===== Cell: smoke tests =====
results = []


def check(name, fn):
    try:
        results.append((name, "ok", str(fn())[:80]))
    except Exception as e:
        results.append((name, "RAISED", f"{type(e).__name__}: {e}"))


def t_escape():
    for bad in ["/etc/passwd", "/home/princ3kr/../../etc/passwd"]:
        try:
            safe_resolve(bad); raise AssertionError(f"LEAK {bad}")
        except PermissionError:
            pass
    return "blocked absolute + traversal"


def t_allowlist():
    try:
        call_tool("delete_everything"); raise AssertionError("breach")
    except ValueError:
        return "unknown tool refused"


def t_missing():
    r = read_file("/home/princ3kr/nope_12345.txt")
    assert r["status"] == "not_found", r
    return r["status"]


def t_binary():
    import tempfile, os, shutil
    d = tempfile.mkdtemp(dir=ROOTS[0], prefix="_smoke_")
    try:
        fd_, tmp = tempfile.mkstemp(suffix=".dat", dir=d)
        os.write(fd_, b"\x00\x01\x02" * 100); os.close(fd_)
        r = read_file(tmp)
        assert r["status"] == "binary", r
        return r["status"]
    finally:
        shutil.rmtree(d, ignore_errors=True)


def t_outside():
    import tempfile, os
    fd_, tmp = tempfile.mkstemp(suffix=".dat", dir="/tmp")
    os.write(fd_, b"hi"); os.close(fd_)
    try:
        r = read_file(tmp)
        assert r["status"] == "blocked", r
        return r["status"]
    finally:
        os.unlink(tmp)


def t_caps():
    r = read_file("/home/princ3kr/Desktop/learnings/glinerTest.ipynb")
    assert len(r.get("text", "")) <= MAX_FILE_BYTES * 1.2
    return f"{r['n_lines']}/{r['total_lines']} lines"


def t_grep():
    hits = search_content("ipynb", root="/home/princ3kr/Desktop")
    assert hits and hits[0]["line_no"] > 0
    return f"{len(hits)} hit(s), first L{hits[0]['line_no']}"


def t_notfound():
    import tempfile, shutil
    d = tempfile.mkdtemp(dir=ROOTS[0], prefix="_empty_")
    try:
        # built at RUNTIME so this notebook's own source cannot match it
        needle = "zzz" + "qqx" + "nomatch" + "99871"
        r = run_agent(needle, root=d)
        assert r["status"] == "not_found" and r["confidence"] == 0.0, r
        assert "not_found" in build_prompt_bundle(r)
        return "graceful not_found"
    finally:
        shutil.rmtree(d, ignore_errors=True)


def t_root_override():
    try:
        find_files("*.py", root="/etc"); raise AssertionError("leak")
    except PermissionError:
        return "root= blocked"


# --- v2: exhaustive similarity recall
def t_similarity():
    import tempfile, shutil, pathlib
    d = pathlib.Path(tempfile.mkdtemp(dir=ROOTS[0], prefix="_sim_"))
    try:
        for n in ["utils.py", "utils_test.py", "my_utils.py", "util.py", "unrelated.py"]:
            (d / n).write_text("x = 1\n")
        got = {f["path"].name: f["kind"] for f in find_files("utils.py", root=d)}
        must = {"utils.py", "utils_test.py", "my_utils.py", "util.py"}
        assert must <= set(got), f"missed: {must - set(got)}"
        assert "unrelated.py" not in got, f"false positive: {got}"
        assert got["utils.py"] == "exact" and got["util.py"] == "fuzzy", got
        return "4/4 recalled, 1 rejected, kinds correct"
    finally:
        shutil.rmtree(d, ignore_errors=True)


# --- v2: full loop -> natural language, with the transport stubbed
def t_llm_loop():
    global LLM_API_KEY                       # call_llm short-circuits on an empty key,
    orig_key = LLM_API_KEY                   # so the stub must supply one too
    LLM_API_KEY = "sk-test-stub"
    class Resp:
        status_code = 200
        def json(self):
            return {"model": "stub-llm",
                    "choices": [{"message": {"content":
                        "safe_resolve is defined in glinerTest.ipynb and enforces the jail."}}],
                    "usage": {"total_tokens": 42}}
    orig_post = requests.post
    captured = {}

    def fake_post(url, headers=None, json=None, timeout=None):
        captured.update({"url": url, "headers": headers, "payload": json})
        return Resp()

    requests.post = fake_post
    try:
        out = synthesize(run_agent("where is safe_resolve defined?"))
        assert out is not None, "synthesize returned None"
        assert "safe_resolve" in out["text"], out
        assert out["usage"]["total_tokens"] == 42, out
        # the outgoing request must be a well-formed OpenAI payload
        assert captured["url"].endswith("/chat/completions"), captured["url"]
        assert captured["payload"]["model"] and captured["payload"]["temperature"] == 0.0
        assert captured["payload"]["messages"][0]["role"] == "system"
        assert "Authorization" in captured["headers"]
        return f"natural language via stub ({out['model']})"
    finally:
        requests.post = orig_post
        LLM_API_KEY = orig_key


def t_llm_offline():
    if LLM_API_KEY:
        return "skipped (key present)"
    assert llm_status()["configured"] is False
    assert synthesize({"status": "not_found", "query": "x", "searches": [],
                       "evidence": []}) is None
    return "degrades, no crash"


for nm, fn in [("jail: escapes blocked", t_escape),
               ("allowlist fails closed", t_allowlist),
               ("missing file", t_missing),
               ("binary detection", t_binary),
               ("outside jail refused", t_outside),
               ("size caps", t_caps),
               ("grep + line numbers", t_grep),
               ("graceful not_found", t_notfound),
               ("root override blocked", t_root_override),
               ("exhaustive similarity", t_similarity),
               ("LLM loop -> prose", t_llm_loop),
               ("LLM offline degrades", t_llm_offline)]:
    check(nm, fn)

print(f"{'test':<26} {'result':<8} detail")
print("-" * 78)
for nm, st_, d_ in results:
    print(f"{nm:<26} {st_:<8} {d_}{'' if st_ == 'ok' else '  <-- '}")
bad = [r for r in results if r[1] not in ("ok", "skipped")]
print("-" * 78)
print(f"{len(results) - len(bad)}/{len(results)} passed")
````

## What's next

This milestone answers: *can the harness find exhaustively, read safely, and answer in
prose?* Still deliberately absent:

| Next | Why |
|---|---|
| **result contract** as a dataclass | `status, summary, evidence, confidence` is already the shape; formalise it before agent #2 depends on it |
| `planner` | no tools; its output schema becomes the future executor's *input schema* |
| `webSearch` | check `importlib`/`pip show` locally **before** going online; treat fetched text as untrusted data |
| streaming the LLM answer | `stream=True` — latency, not correctness |
| embeddings for recall | lexical now; only add when fuzzy recall actually misses |
| GLiNER → entity args | `gliner` + `torch` installed, so cheap to try — but GLiNER finds **what the user mentioned**, not what they want |
| DSPy router | replaces the hardcoded `find vs grep`; score on 30–50 labelled queries |

**Known limitations, on purpose**
1. **`find_files` has no global ranking across multiple named files.** If you ask for
   `a.py and b.py`, each is expanded separately and results are concatenated, not merged
   by similarity. Fix: one ranking pass over the union.
2. **Exhaustive name search never falls back to content search.** If a filename was named
   and found, symbols inside it are not additionally grepped.
3. **`search_content` has no file-size guard** — this grep has no `--max-filesize`, so a
   large text file in an unpruned directory is fully scanned. `SKIP_DIRS` compensates;
   the real fix is a `find -size` walk or an index.
4. **The LLM answer is unverified.** It can misread the evidence. A `verifier` agent that
   re-checks each citation against the file is the natural next guard — and cheap,
   because citations are already structured.
